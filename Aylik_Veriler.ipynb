4)TotalDowntime Min. sütununun dağılımını box plot ile görselleştirme

```python
plt.figure(figsize=(12, 2.5))  # Yüksekliği azalt, genişliği artır
sns.boxplot(
    x=data['TotalDowntimeMinute'],
    color='darkorange',
    linewidth=1.5,
    fliersize=3,  # outlier noktalarının boyutu
    boxprops=dict(edgecolor='black'),
    medianprops=dict(color='black'),
    whiskerprops=dict(color='black'),
    capprops=dict(color='black')
)

plt.title('TotalDowntimeMinute Dağılımı (Box Plot)', fontsize=14)
plt.xlabel('TotalDowntimeMinute', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

🔸 Kutu ve Bıyıklar (Whiskers):
Ortadaki kutu, 25. yüzdelik (Q1) ile 75. yüzdelik (Q3) değerler arasındaki verileri kapsar.

Kutunun içindeki siyah çizgi, medyanı (ortalama değil!) gösterir.

Kutunun dışındaki çizgiler (whiskers), uç değerler hariç maksimum ve minimum değer aralığını gösterir.

🔸 Aykırı Değerler (Outliers):
Kutunun dışında kalan tekil noktalar, standart istatistik kurallarına göre aykırı değerlerdir.

Bu grafikte çok sayıda aykırı değer var. Özellikle:

2000 dakikadan büyük duruş süreleri oldukça fazla.

4000 dakikayı, hatta 6000 dakikayı geçen ekstrem vakalar mevcut.


Bu grafik bize şunu gösteriyor:

Bazı tekil olaylar olağan dışı şekilde uzun süre durmuş!

```python
q1 = data['TotalDowntimeMinute'].quantile(0.25)
q3 = data['TotalDowntimeMinute'].quantile(0.75)
iqr = q3 - q1
upper_bound = q3 + 1.5 * iqr

filtered_data = data[data['TotalDowntimeMinute'] <= upper_bound]

plt.figure(figsize=(10, 2.5))
sns.boxplot(x=filtered_data['TotalDowntimeMinute'], color='seagreen')
plt.title('Outlier Olmayan Verilerle Box Plot')
plt.xlabel('TotalDowntimeMinute')
plt.tight_layout()
plt.show()
```

🔸 Duruş Süresi Dağılımı:
Medyan (orta çizgi) yaklaşık 20-25 dakika civarında.

Verilerin %50’lik kısmı, yaklaşık 10 ile 80 dakika arasında yer alıyor. Bu da tipik bir duruş süresinin bu aralıkta olduğunu gösteriyor.

Whisker (bıyık) uç noktası yaklaşık 140 dakika. Yani 140 dakikaya kadar olan veriler normal sınırlar içinde kabul edilmiş.

🔸 Hâlâ Tekil Değerler Mevcut:
140 dakikadan sonra hâlâ birkaç adet daha küçük çaplı outlier görünüyor (200-250-350 dk gibi). Ancak bu değerler, ilk grafikteki 6000 dakika gibi ekstrem uçlardan çok daha az sayıda ve daha düşük seviyede.

Duruşların büyük çoğunluğu kısa süreli. Bu da operasyonların genellikle çabuk toparlandığını gösteriyor.

Ancak, 200 dakika ve üzeri duruşlar hâlâ az da olsa var.

```python
# TotalDowntime Min. sütununun dağılımını box plot ile görselleştirme
plt.figure(figsize=(10,6))
sns.boxplot(x=data['TotalDowntimeMinute'], color='orange')
plt.title('TotalDowntimeMinuteDağılımı (Box Plot)')
plt.xlabel('TotalDowntimeMinute')
plt.show()
```

🔸 Medyan ve Kutunun Yapısı:
Kutu (box), verilerin orta %50'lik kısmını (Q1 ile Q3 arası) temsil eder.

Bu kutu, duruş sürelerinin büyük bölümünün düşük olduğunu gösteriyor (yaklaşık 0–300 dakika arası).

🔸 Whisker (bıyık) sınırı:
Sağdaki dikey çizgi, normal sınırlar içindeki en büyük duruş süresini gösteriyor. Bu da yaklaşık 500-600 dakika civarı.

🔸 Aykırı Değerler (Outliers):
600 dakikadan sonra başlayan yüzlerce nokta, outlier’dır (aykırı değer).

Bunlar:

1000, 2000, 3000… 6000 dakika gibi uçuk seviyelerdeki duruşları temsil eder.

Bu değerler, sistemde bazı işlerin normalden çok daha uzun süre durduğunu açıkça ortaya koyuyor.

En uç noktada yaklaşık 6300 dakikalık bir duruş süresi var — bu 4.3 gün eder.

Adım 18: İki Kategorik Değişken Arasındaki İlişkiyi İncelemek (pd.crosstab)

```python
# İki kategorik değişken arasındaki ilişki
contingency_table = pd.crosstab(data['Workcenter'], data['DownTime_Reason'])
print(contingency_table)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

import shap
import warnings
warnings.filterwarnings("ignore")
```

```python
X = df[[
    'Years',
    'Months',
    'TotalNofOccured',
    'Quarter',
    'Rolling3_TotalNofOccured',
    'Rolling3_DowntimeAvg',
    'Workcenter_Encoded',
    'DownTime_Reason_Encoded'
]]
y = df['TotalDowntimeMinute']
```

```python
models = {
    "LinearRegression": LinearRegression(),
    "Ridge": Ridge(),
    "Lasso": Lasso(),
    "RandomForest": RandomForestRegressor(random_state=42),
    "GradientBoosting": GradientBoostingRegressor(random_state=42),
    "KNN": KNeighborsRegressor(),
    "SVR": SVR(),
    "XGBoost": XGBRegressor(random_state=42),
    "LightGBM": LGBMRegressor(random_state=42)
}

results = {}

for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)

    results[name] = {
        "R2": r2_score(y_test, y_pred),
        "MAE": mean_absolute_error(y_test, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, y_pred)),
        "Residuals": y_test - y_pred
    }

    print(f"\n🔹 {name}")
    print("R2:", results[name]['R2'])
    print("MAE:", results[name]['MAE'])
    print("RMSE:", results[name]['RMSE'])
```
